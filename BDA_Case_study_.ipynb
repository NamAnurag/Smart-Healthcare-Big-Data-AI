## 5. Feature-Target Relationship Analysis

````python
fall_corr = (
    df.select_dtypes(include=np.number)
      .corr()["fall_detected"]
      .sort_values(ascending=False)
)

print(fall_corr)
````

````python
from sklearn.feature_selection import mutual_info_classif

features = [
    "heart_rate",
    "spo2_level",
    "ecg_signal",
    "respiration_rate",
    "body_temperature",
    "blood_pressure_sys",
    "blood_pressure_dia",
    "blood_glucose",
    "eeg_alpha_power",
    "eeg_beta_power",
    "emg_signal_strength",
    "step_count",
    "ambient_temperature",
    "stress_level_index"
]

X_mi = df[features]
y_mi = df["fall_detected"]

mi_scores = mutual_info_classif(
    X_mi,
    y_mi,
    random_state=42
)

mi_df = pd.DataFrame({
    "Feature": features,
    "Mutual_Information": mi_scores
}).sort_values(
    "Mutual_Information",
    ascending=False
)

mi_df
````

````python
plt.figure(figsize=(10, 6))

plt.barh(
    mi_df["Feature"],
    mi_df["Mutual_Information"]
)

plt.xlabel("Mutual Information")
plt.ylabel("Feature")
plt.title("Feature Relevance for Fall Detection")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
````

````python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

X = df[features]
y = df["fall_detected"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
````

````python
baseline_accuracy = (y_test == 0).mean()

print("Majority-class baseline accuracy:",
      round(baseline_accuracy, 4))
````

````python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Fall", "Fall"]
)

disp.plot()

plt.title("Decision Tree Confusion Matrix")
plt.tight_layout()
plt.show()
````

## 6. Machine Learning: Activity Classification

````python
features_activity = [
    "heart_rate",
    "spo2_level",
    "ecg_signal",
    "respiration_rate",
    "body_temperature",
    "blood_pressure_sys",
    "blood_pressure_dia",
    "blood_glucose",
    "eeg_alpha_power",
    "eeg_beta_power",
    "emg_signal_strength",
    "step_count",
    "ambient_temperature",
    "stress_level_index"
]
````

````python
X = df[features_activity]
y = df["activity_type"]
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
````

````python
from sklearn.tree import DecisionTreeClassifier

activity_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=8
)

activity_model.fit(X_train, y_train)

y_pred_activity = activity_model.predict(X_test)
````

````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Accuracy :", round(accuracy_score(y_test, y_pred_activity), 4))
print("Precision:", round(
    precision_score(y_test, y_pred_activity, average="macro"),
    4
))
print("Recall   :", round(
    recall_score(y_test, y_pred_activity, average="macro"),
    4
))
print("F1 Score :", round(
    f1_score(y_test, y_pred_activity, average="macro"),
    4
))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_activity))
````

````python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_activity,
    labels=["resting", "walking", "running", "sleeping"]
)

plt.title("Decision Tree Confusion Matrix - Activity Classification")
plt.tight_layout()
plt.show()
````

````python
feature_importance = pd.DataFrame({
    "Feature": features_activity,
    "Importance": activity_model.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

feature_importance
````

````python
plt.figure(figsize=(10, 6))

plt.barh(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Decision Tree Feature Importance")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
````

````python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("Random Forest Accuracy:",
      round(accuracy_score(y_test, rf_pred), 4))

print("Random Forest Macro F1:",
      round(f1_score(y_test, rf_pred, average="macro"), 4))
````

````python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = cross_val_score(
    rf_model,
    X,
    y,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

print("Cross-validation accuracies:")
print(cv_scores)

print("\nMean CV Accuracy:",
      round(cv_scores.mean(), 4))

print("Standard Deviation:",
      round(cv_scores.std(), 4))
````

````python
from sklearn.utils import shuffle

y_shuffled = shuffle(
    y,
    random_state=42
).reset_index(drop=True)

X_reset = X.reset_index(drop=True)

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_reset,
    y_shuffled,
    test_size=0.20,
    random_state=42,
    stratify=y_shuffled
)

rf_sanity = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_sanity.fit(X_train_s, y_train_s)

sanity_pred = rf_sanity.predict(X_test_s)

print(
    "Accuracy with shuffled labels:",
    round(accuracy_score(y_test_s, sanity_pred), 4)
)
````

## 8. Time-Based Model Validation

````python
# Sort chronologically
time_df = df.sort_values("timestamp").reset_index(drop=True)

# 80% earliest records for training
# 20% latest records for testing
split_index = int(len(time_df) * 0.80)

train_time = time_df.iloc[:split_index]
test_time = time_df.iloc[split_index:]

X_train_time = train_time[features_activity]
y_train_time = train_time["activity_type"]

X_test_time = test_time[features_activity]
y_test_time = test_time["activity_type"]

print("Training records:", len(X_train_time))
print("Testing records:", len(X_test_time))

print("\nTraining period:")
print(train_time["timestamp"].min(), "to", train_time["timestamp"].max())

print("\nTesting period:")
print(test_time["timestamp"].min(), "to", test_time["timestamp"].max())
````

````python
rf_time = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_time.fit(X_train_time, y_train_time)

time_pred = rf_time.predict(X_test_time)
````

````python
print("Time-Based Accuracy:",
      round(accuracy_score(y_test_time, time_pred), 4))

print("Time-Based Macro F1:",
      round(
          f1_score(
              y_test_time,
              time_pred,
              average="macro"
          ),
          4
      ))

print("\nClassification Report:\n")
print(classification_report(y_test_time, time_pred))
````

````python
import pandas as pd

df = pd.read_csv("healthcare_iot_data.csv")

print("Shape:", df.shape)
````

````python
risk_features = [
    "heart_rate",
    "spo2_level",
    "respiration_rate",
    "body_temperature",
    "blood_pressure_sys",
    "blood_pressure_dia",
    "blood_glucose",
    "stress_level_index",
    "step_count"
]

X_risk = df[risk_features].copy()

print("Features used:")
print(risk_features)
print("\nShape:", X_risk.shape)
````

````python
k_values = range(2, 7)

inertias = []
silhouette_scores = []

for k in k_values:
    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = km.fit_predict(X_risk_scaled)

    inertias.append(km.inertia_)
    silhouette_scores.append(
        silhouette_score(X_risk_scaled, labels)
    )

results = pd.DataFrame({
    "k": list(k_values),
    "inertia": inertias,
    "silhouette_score": silhouette_scores
})

results
````

## 12. Interactive Healthcare Monitoring Profile

The trained KMeans model assigns a new set of sensor readings to one of the
data-derived healthcare monitoring profiles learned from the dataset.

The output represents a statistical profile only and is not a medical diagnosis.

````python
def predict_monitoring_profile(
    heart_rate,
    spo2_level,
    blood_glucose,
    stress_level_index,
    step_count
):
    patient_values = pd.DataFrame([{
        "heart_rate": heart_rate,
        "spo2_level": spo2_level,
        "blood_glucose": blood_glucose,
        "stress_level_index": stress_level_index,
        "step_count": step_count
    }])

    # Ensure the same feature order used during training
    patient_values = patient_values[selected_features]

    # Apply the same scaler used during model training
    patient_scaled = scaler_selected.transform(patient_values)

    # Predict cluster
    cluster = int(
        final_kmeans.predict(patient_scaled)[0]
    )

    # Distance from each cluster centroid
    distances = final_kmeans.transform(patient_scaled)[0]

    nearest_distance = float(distances[cluster])

    print("Healthcare Monitoring Result")
    print("-" * 35)
    print("Assigned profile:", cluster)
    print(
        "Distance from profile centroid:",
        f"{nearest_distance:.3f}"
    )
    print()
    print(
        "Interpretation: This record is assigned to "
        "a data-derived monitoring profile learned "
        "from the training dataset."
    )
    print(
        "This is not a medical diagnosis or clinical risk assessment."
    )

    return cluster
````

````python
def compare_with_profile(
    heart_rate,
    spo2_level,
    blood_glucose,
    stress_level_index,
    step_count
):
    patient = pd.DataFrame([{
        "heart_rate": heart_rate,
        "spo2_level": spo2_level,
        "blood_glucose": blood_glucose,
        "stress_level_index": stress_level_index,
        "step_count": step_count
    }])

    patient = patient[selected_features]

    patient_scaled = scaler_selected.transform(patient)

    cluster = int(
        final_kmeans.predict(patient_scaled)[0]
    )

    profile = final_profile_display.loc[cluster]

    comparison = pd.DataFrame({
        "Feature": selected_features,
        "Input": [
            heart_rate,
            spo2_level,
            blood_glucose,
            stress_level_index,
            step_count
        ],
        "Profile Average": [
            profile["heart_rate"],
            profile["spo2_level"],
            profile["blood_glucose"],
            profile["stress_level_index"],
            profile["step_count"]
        ]
    })

    comparison["Difference"] = (
        comparison["Input"] -
        comparison["Profile Average"]
    ).round(2)

    print(f"Assigned Monitoring Profile: {cluster}")
    print()
    print(comparison.to_string(index=False))

    print(
        "\nNote: Differences describe statistical similarity "
        "to the learned profile and are not medical assessments."
    )

    return comparison
````

````python
# ==============================================================
#      SMART HEALTHCARE AI — FINAL TEACHER DEMO DASHBOARD
# ==============================================================
#
# COMPLETE PIPELINE
#
# Manual Sensor Input
#        ↓
#      K-Means
#        ↓
# Monitoring Profile
#
# Held-out Test Record
#        ↓
# Random Forest
#        ↓
# Activity Classification
#
# Monitoring Profile
#        ↓
# Generative AI
#        ↓
# Patient-Friendly Explanation
#
# ==============================================================


# ==============================================================
# 1. INSTALL REQUIRED PACKAGES
# ==============================================================

!pip -q install -U google-genai ipywidgets


# ==============================================================
# 2. IMPORTS
# ==============================================================

import os
import re
import html
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    silhouette_score
)

import ipywidgets as widgets

from IPython.display import (
    display,
    HTML,
    clear_output
)


# ==============================================================
# 3. LOAD DATA
# ==============================================================

FILE = "healthcare_iot_data.csv"

if not os.path.exists(FILE):

    raise FileNotFoundError(
        f"'{FILE}' was not found.\n"
        "Upload healthcare_iot_data.csv to the Colab file panel."
    )

df = pd.read_csv(FILE)

if "timestamp" in df.columns:

    df["timestamp"] = pd.to_datetime(
        df["timestamp"],
        errors="coerce"
    )


# ==============================================================
# 4. VERIFY REQUIRED COLUMNS
# ==============================================================

required_columns = [

    # K-Means
    "heart_rate",
    "spo2_level",
    "blood_glucose",
    "stress_level_index",
    "step_count",

    # Random Forest
    "eeg_beta_power",
    "ambient_temperature",
    "emg_signal_strength",
    "body_temperature",
    "ecg_signal",
    "eeg_alpha_power",
    "respiration_rate",
    "blood_pressure_dia",
    "blood_pressure_sys",

    # Target
    "activity_type"
]

missing = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing:

    raise ValueError(
        "Missing required columns:\n"
        + ", ".join(missing)
    )


# ==============================================================
# 5. K-MEANS MODEL
# ==============================================================

kmeans_features = [
    "heart_rate",
    "spo2_level",
    "blood_glucose",
    "stress_level_index",
    "step_count"
]

kmeans_scaler = StandardScaler()

X_kmeans = kmeans_scaler.fit_transform(
    df[kmeans_features]
)

kmeans_model = KMeans(
    n_clusters=12,
    random_state=42,
    n_init=30
)

df["monitoring_profile"] = (
    kmeans_model.fit_predict(
        X_kmeans
    )
)

kmeans_silhouette = silhouette_score(
    X_kmeans,
    df["monitoring_profile"]
)

profile_summary = (
    df
    .groupby("monitoring_profile")[kmeans_features]
    .mean()
    .round(2)
)

profile_counts = (
    df["monitoring_profile"]
    .value_counts()
    .sort_index()
)


# ==============================================================
# 6. RANDOM FOREST MODEL
# ==============================================================
#
# IMPORTANT:
# We split INDICES so the dashboard can later select an actual
# held-out test record.
# ==============================================================

rf_features = [

    "stress_level_index",
    "step_count",
    "blood_glucose",
    "eeg_beta_power",
    "heart_rate",
    "ambient_temperature",
    "emg_signal_strength",
    "body_temperature",
    "ecg_signal",
    "spo2_level",
    "eeg_alpha_power",
    "respiration_rate",
    "blood_pressure_dia",
    "blood_pressure_sys"
]

train_indices, test_indices = train_test_split(
    df.index.to_numpy(),
    test_size=0.20,
    random_state=42,
    stratify=df["activity_type"]
)

train_indices = np.sort(train_indices)
test_indices = np.sort(test_indices)

X_train = df.loc[
    train_indices,
    rf_features
]

X_test = df.loc[
    test_indices,
    rf_features
]

y_train = df.loc[
    train_indices,
    "activity_type"
]

y_test = df.loc[
    test_indices,
    "activity_type"
]

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train,
    y_train
)

rf_predictions = rf_model.predict(
    X_test
)

rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)

rf_macro_f1 = f1_score(
    y_test,
    rf_predictions,
    average="macro"
)


# ==============================================================
# 7. HELPER FUNCTIONS
# ==============================================================

FONT = (
    "-apple-system,"
    "'Segoe UI',"
    "Roboto,"
    "Helvetica,"
    "Arial,"
    "sans-serif"
)


def make_html(
    content,
    width="100%",
    height=None
):

    kwargs = {
        "width": width
    }

    if height is not None:

        kwargs["height"] = height

    return widgets.HTML(

        value=content,

        layout=widgets.Layout(
            **kwargs
        )
    )


def card(
    content,
    accent="#2563eb"
):

    return f"""
    <div style="
        width:100%;
        box-sizing:border-box;
        background:#ffffff;
        border:1px solid #d1d5db;
        border-left:6px solid {accent};
        border-radius:12px;
        padding:18px 20px;
        margin:10px 0;
        font-family:{FONT};
        color:#111827;
    ">

        {content}

    </div>
    """


def section_title(
    title,
    subtitle=""
):

    result = f"""
    <div style="
        font-size:20px;
        font-weight:800;
        color:#111827;
        font-family:{FONT};
    ">
        {title}
    </div>
    """

    if subtitle:

        result += f"""
        <div style="
            margin-top:7px;
            font-size:13px;
            line-height:1.6;
            color:#374151;
            font-family:{FONT};
        ">
            {subtitle}
        </div>
        """

    return result


def metric_card(
    value,
    label,
    value_color="#2563eb"
):

    return f"""
    <div style="
        width:155px;
        min-width:155px;
        height:80px;
        box-sizing:border-box;
        background:#f8fafc;
        border:1px solid #d1d5db;
        border-radius:10px;
        padding:12px 8px;
        text-align:center;
        font-family:{FONT};
    ">

        <div style="
            color:{value_color};
            font-size:23px;
            font-weight:800;
        ">
            {value}
        </div>

        <div style="
            color:#4b5563;
            font-size:11px;
            font-weight:700;
            margin-top:7px;
            line-height:1.2;
        ">
            {label}
        </div>

    </div>
    """


def info_table(
    rows,
    headers=("Feature", "Value")
):

    header_html = ""

    for header in headers:

        header_html += f"""
        <th style="
            text-align:left;
            padding:9px 11px;
            background:#f3f4f6;
            color:#111827;
            font-size:11px;
            font-weight:800;
            border-bottom:1px solid #d1d5db;
        ">
            {header}
        </th>
        """

    body_html = ""

    for row in rows:

        cells = ""

        for cell in row:

            cells += f"""
            <td style="
                padding:8px 11px;
                color:#374151;
                font-size:13px;
                border-bottom:1px solid #e5e7eb;
            ">
                {cell}
            </td>
            """

        body_html += f"<tr>{cells}</tr>"

    return f"""
    <table style="
        width:100%;
        border-collapse:collapse;
        margin-top:10px;
        font-family:{FONT};
        background:#ffffff;
    ">

        <thead>
            <tr>
                {header_html}
            </tr>
        </thead>

        <tbody>
            {body_html}
        </tbody>

    </table>
    """


def label_widget(
    text,
    width="135px"
):

    return widgets.HTML(

        value=f"""
        <div style="
            width:{width};
            box-sizing:border-box;
            padding:8px 10px;
            border:1px solid #d1d5db;
            border-radius:7px;
            background:#f9fafb;
            color:#111827 !important;
            font-family:{FONT};
            font-size:13px;
            font-weight:700;
            text-align:left;
        ">
            {text}
        </div>
        """,

        layout=widgets.Layout(
            width=width,
            height="38px",
            margin="0 8px 0 0"
        )
    )


def input_row(
    label,
    widget
):

    return widgets.HBox(
        [
            label_widget(label),
            widget
        ],
        layout=widgets.Layout(
            width="330px",
            height="42px",
            align_items="center",
            margin="3px 0"
        )
    )


def styled_float(
    value
):

    return widgets.FloatText(

        value=value,

        step=0.1,

        layout=widgets.Layout(
            width="150px",
            height="36px"
        )
    )


def styled_int(
    value
):

    return widgets.IntText(

        value=value,

        step=1,

        layout=widgets.Layout(
            width="150px",
            height="36px"
        )
    )


def action_button(
    text,
    color
):

    return widgets.Button(

        description=text,

        button_style="",

        layout=widgets.Layout(
            width="270px",
            height="42px",
            margin="10px 0"
        ),

        style=widgets.ButtonStyle(
            button_color=color,
            font_weight="bold"
        )
    )


# ==============================================================
# 8. HEADER
# ==============================================================

display(
    HTML(
        f"""
        <div style="
            width:100%;
            box-sizing:border-box;
            background:#0f172a;
            border-radius:14px;
            padding:23px;
            margin:10px 0 14px 0;
            text-align:center;
            font-family:{FONT};
        ">

            <div style="
                color:#ffffff !important;
                font-size:28px;
                font-weight:800;
                letter-spacing:1px;
            ">
                SMART HEALTHCARE AI
            </div>

            <div style="
                color:#cbd5e1 !important;
                margin-top:8px;
                font-size:13px;
                line-height:1.5;
            ">
                IoT Sensor Analytics
                &nbsp;•&nbsp;
                K-Means Monitoring Profiles
                &nbsp;•&nbsp;
                Random Forest Activity Classification
                &nbsp;•&nbsp;
                Generative AI
            </div>

        </div>
        """
    )
)


# ==============================================================
# 9. MODEL PERFORMANCE
# ==============================================================

performance_html = card(

    section_title(
        "Model Performance",
        "Measured using the trained models and the held-out "
        "test split for Random Forest."
    )

    +

    f"""
    <div style="
        display:flex;
        flex-wrap:wrap;
        gap:12px;
        margin-top:16px;
    ">

        {metric_card(
            f"{rf_accuracy:.4f}",
            "RF Test Accuracy",
            "#15803d"
        )}

        {metric_card(
            f"{rf_macro_f1:.4f}",
            "RF Macro F1",
            "#15803d"
        )}

        {metric_card(
            "12",
            "K-Means Profiles",
            "#2563eb"
        )}

        {metric_card(
            f"{kmeans_silhouette:.4f}",
            "K-Means Silhouette",
            "#7c3aed"
        )}

    </div>
    """,

    accent="#16a34a"
)

display(
    HTML(performance_html)
)


# ==============================================================
# 10. SECTION 1 — K-MEANS
# ==============================================================

display(
    HTML(
        card(
            section_title(
                "1 · Live Monitoring Profile — K-Means",
                "Enter five monitoring values. The trained K-Means "
                "model assigns the observation to one of 12 "
                "data-derived monitoring profiles."
            ),
            accent="#2563eb"
        )
    )
)


# --------------------------------------------------------------
# Input widgets
# --------------------------------------------------------------

heart_rate_input = styled_float(90)

spo2_input = styled_float(96)

glucose_input = styled_float(130)

stress_input = styled_float(50)

steps_input = styled_int(900)


kmeans_input_layout = widgets.VBox(

    [

        widgets.HBox(
            [
                input_row(
                    "Heart Rate",
                    heart_rate_input
                ),

                input_row(
                    "SpO₂",
                    spo2_input
                )
            ]
        ),

        widgets.HBox(
            [
                input_row(
                    "Blood Glucose",
                    glucose_input
                ),

                input_row(
                    "Stress Index",
                    stress_input
                )
            ]
        ),

        input_row(
            "Step Count",
            steps_input
        )

    ],

    layout=widgets.Layout(
        width="100%",
        margin="5px 0"
    )
)


predict_profile_button = action_button(
    "Predict Monitoring Profile",
    "#2563eb"
)

profile_output = widgets.Output()


display(
    widgets.VBox(
        [
            kmeans_input_layout,
            predict_profile_button,
            profile_output
        ],
        layout=widgets.Layout(
            width="100%"
        )
    )
)


# --------------------------------------------------------------
# K-Means prediction
# --------------------------------------------------------------

def predict_profile(_):

    with profile_output:

        clear_output(
            wait=True
        )

        try:

            values = np.array([
                [
                    float(heart_rate_input.value),
                    float(spo2_input.value),
                    float(glucose_input.value),
                    float(stress_input.value),
                    float(steps_input.value)
                ]
            ])

            scaled_values = (
                kmeans_scaler.transform(
                    values
                )
            )

            profile = int(
                kmeans_model.predict(
                    scaled_values
                )[0]
            )

            centroid = (
                kmeans_model
                .cluster_centers_[profile]
            )

            distance = float(
                np.linalg.norm(
                    scaled_values[0]
                    - centroid
                )
            )

            averages = (
                profile_summary
                .loc[profile]
            )

            record_count = int(
                profile_counts
                .loc[profile]
            )

            percentage = (
                record_count
                / len(df)
                * 100
            )

            rows = [

                (
                    "Heart Rate",
                    f"{averages['heart_rate']:.2f}"
                ),

                (
                    "SpO₂",
                    f"{averages['spo2_level']:.2f}"
                ),

                (
                    "Blood Glucose",
                    f"{averages['blood_glucose']:.2f}"
                ),

                (
                    "Stress Index",
                    f"{averages['stress_level_index']:.2f}"
                ),

                (
                    "Step Count",
                    f"{averages['step_count']:.2f}"
                )

            ]

            body = f"""

            <div style="
                color:#374151;
                font-size:14px;
                line-height:1.7;
            ">

                <div>
                    <b style="color:#111827;">
                        Cluster distance:
                    </b>
                    {distance:.3f}
                </div>

                <div>
                    <b style="color:#111827;">
                        Records in profile:
                    </b>
                    {record_count:,}
                    ({percentage:.2f}% of dataset)
                </div>

                <div style="
                    margin-top:14px;
                    color:#111827;
                    font-weight:800;
                    font-size:15px;
                ">
                    Profile {profile} Averages
                </div>

                {info_table(
                    rows,
                    headers=(
                        "Feature",
                        "Profile Average"
                    )
                )}

                <div style="
                    background:#fffbeb;
                    border:1px solid #fde68a;
                    border-radius:8px;
                    padding:10px 12px;
                    margin-top:14px;
                    color:#78350f;
                    font-size:12.5px;
                    line-height:1.6;
                ">
                    This is a statistical monitoring profile
                    produced by K-Means clustering. It is not
                    a medical diagnosis or clinical risk category.
                </div>

            </div>
            """

            display(
                HTML(
                    card(

                        f"""
                        <div style="
                            color:#2563eb;
                            font-size:28px;
                            font-weight:800;
                            margin-bottom:10px;
                        ">
                            Profile {profile}
                        </div>
                        """

                        + body,

                        accent="#2563eb"
                    )
                )
            )

        except Exception as exc:

            display(
                HTML(
                    card(
                        f"""
                        <div style="
                            color:#b91c1c;
                            font-size:18px;
                            font-weight:800;
                        ">
                            Input Error
                        </div>

                        <div style="
                            color:#374151;
                            margin-top:8px;
                        ">
                            {html.escape(str(exc))}
                        </div>
                        """,

                        accent="#dc2626"
                    )
                )
            )


predict_profile_button.on_click(
    predict_profile
)


# Show initial profile automatically
predict_profile_button.click()


# ==============================================================
# 11. SECTION 2 — RANDOM FOREST
# ==============================================================

display(
    HTML(
        card(
            section_title(
                "2 · Activity Classification — Random Forest",
                "The trained Random Forest uses the complete "
                "14-feature sensor vector. The demo below "
                "evaluates a genuinely held-out test observation."
            )

            +

            """
            <div style="
                background:#eff6ff;
                border:1px solid #bfdbfe;
                border-radius:9px;
                padding:11px 13px;
                margin-top:13px;
                color:#1e3a8a;
                font-size:12.5px;
                line-height:1.6;
            ">
                <b>Why a test record?</b>
                Random Forest was trained on 80% of the data.
                The selector below uses only the remaining
                20% held-out test set for demonstration.
            </div>
            """,

            accent="#16a34a"
        )
    )
)


# --------------------------------------------------------------
# Held-out test record selector
# --------------------------------------------------------------

# Use record 437 as default ONLY if it belongs to the test set.
preferred_original_index = 437

if preferred_original_index in set(
    test_indices
):

    default_test_position = list(
        test_indices
    ).index(
        preferred_original_index
    )

else:

    default_test_position = 0


test_record_slider = widgets.IntSlider(

    value=default_test_position,

    min=0,

    max=len(test_indices) - 1,

    step=1,

    continuous_update=False,

    layout=widgets.Layout(
        width="500px",
        height="35px"
    )
)


test_record_label = label_widget(
    "Held-out Test Record",
    width="150px"
)


test_record_row = widgets.HBox(

    [
        test_record_label,
        test_record_slider
    ],

    layout=widgets.Layout(
        width="100%",
        align_items="center",
        margin="8px 0"
    )
)


predict_activity_button = action_button(
    "Predict Activity",
    "#16a34a"
)

activity_output = widgets.Output()


display(
    widgets.VBox(
        [
            test_record_row,
            predict_activity_button,
            activity_output
        ],
        layout=widgets.Layout(
            width="100%"
        )
    )
)


# Clear stale output when slider changes
def clear_activity_output(change):

    with activity_output:

        clear_output(
            wait=True
        )


test_record_slider.observe(
    clear_activity_output,
    names="value"
)


# --------------------------------------------------------------
# Random Forest prediction
# --------------------------------------------------------------

def predict_activity(_):

    with activity_output:

        clear_output(
            wait=True
        )

        try:

            test_position = int(
                test_record_slider.value
            )

            actual_df_index = int(
                test_indices[
                    test_position
                ]
            )

            row = df.loc[
                [actual_df_index]
            ]

            X_one = row[
                rf_features
            ]

            predicted_activity = str(
                rf_model.predict(
                    X_one
                )[0]
            )

            actual_activity = str(
                row[
                    "activity_type"
                ].iloc[0]
            )

            probabilities = (
                rf_model.predict_proba(
                    X_one
                )[0]
            )

            probability_map = dict(
                zip(
                    rf_model.classes_,
                    probabilities
                )
            )

            predicted_probability = (
                probability_map[
                    predicted_activity
                ]
                * 100
            )

            # Five key values for easy teacher viewing
            display_rows = [

                (
                    "Heart Rate",
                    f"{row['heart_rate'].iloc[0]:.2f}"
                ),

                (
                    "SpO₂",
                    f"{row['spo2_level'].iloc[0]:.2f}"
                ),

                (
                    "Blood Glucose",
                    f"{row['blood_glucose'].iloc[0]:.2f}"
                ),

                (
                    "Stress Index",
                    f"{row['stress_level_index'].iloc[0]:.2f}"
                ),

                (
                    "Step Count",
                    f"{row['step_count'].iloc[0]:.0f}"
                )

            ]

            prediction_color = (
                "#15803d"
                if predicted_activity
                == actual_activity
                else "#b45309"
            )

            body = f"""

            <div style="
                color:#374151;
                font-size:14px;
                line-height:1.7;
            ">

                <div>
                    <b style="color:#111827;">
                        Original dataset row:
                    </b>
                    {actual_df_index}
                </div>

                <div>
                    <b style="color:#111827;">
                        Actual dataset label:
                    </b>
                    {actual_activity}
                </div>

                <div>
                    <b style="color:#111827;">
                        Predicted-class probability:
                    </b>
                    {predicted_probability:.2f}%
                </div>

                <div style="
                    color:#111827;
                    font-size:15px;
                    font-weight:800;
                    margin-top:14px;
                ">
                    Key sensor values
                </div>

                {info_table(
                    display_rows,
                    headers=(
                        "Sensor",
                        "Value"
                    )
                )}

                <div style="
                    background:#f0fdf4;
                    border:1px solid #bbf7d0;
                    border-radius:8px;
                    padding:10px 12px;
                    margin-top:14px;
                    color:#166534;
                    font-size:12.5px;
                    line-height:1.6;
                ">
                    The Random Forest prediction uses the
                    complete 14-feature sensor vector from
                    this held-out test observation.
                </div>

            </div>
            """

            display(
                HTML(
                    card(

                        f"""
                        <div style="
                            color:{prediction_color};
                            font-size:28px;
                            font-weight:800;
                            margin-bottom:10px;
                        ">
                            {predicted_activity.upper()}
                        </div>
                        """

                        + body,

                        accent=prediction_color
                    )
                )
            )

        except Exception as exc:

            display(
                HTML(
                    card(

                        f"""
                        <div style="
                            color:#b91c1c;
                            font-size:18px;
                            font-weight:800;
                        ">
                            Prediction Error
                        </div>

                        <div style="
                            color:#374151;
                            margin-top:8px;
                        ">
                            {html.escape(str(exc))}
                        </div>
                        """,

                        accent="#dc2626"
                    )
                )
            )


predict_activity_button.on_click(
    predict_activity
)


# Show initial prediction
predict_activity_button.click()


# ==============================================================
# 12. SECTION 3 — GENERATIVE AI
# ==============================================================

display(
    HTML(
        card(
            section_title(
                "3 · Generative AI Explanation",
                "The K-Means monitoring profile is converted "
                "into a simple, patient-friendly explanation."
            ),
            accent="#b45309"
        )
    )
)


generate_ai_button = action_button(
    "Generate AI Explanation",
    "#f59e0b"
)

genai_output = widgets.Output()


display(
    widgets.VBox(
        [
            generate_ai_button,
            genai_output
        ],
        layout=widgets.Layout(
            width="100%"
        )
    )
)


# ==============================================================
# 13. GENAI MARKDOWN CLEANER
# ==============================================================

def markdown_to_html(
    text
):

    text = str(
        text
    )

    # Escape HTML
    text = (
        text
        .replace("&", "&amp;")
        .replace("<", "&lt;")
        .replace(">", "&gt;")
    )

    # Remove accidentally escaped markdown
    text = text.replace(
        r"\###",
        "###"
    )

    text = text.replace(
        r"\*\*",
        "**"
    )

    lines = text.splitlines()

    rendered = []

    for line in lines:

        line = line.strip()

        if not line:

            rendered.append(
                '<div style="height:7px;"></div>'
            )

            continue

        # Heading
        heading_match = re.match(
            r"^(#{1,6})\s+(.*)$",
            line
        )

        if heading_match:

            level = len(
                heading_match.group(1)
            )

            title = heading_match.group(2)

            title = re.sub(
                r"\*\*(.*?)\*\*",
                r"<b>\1</b>",
                title
            )

            size = {
                1: 19,
                2: 17,
                3: 16
            }.get(
                level,
                15
            )

            rendered.append(
                f"""
                <div style="
                    margin:15px 0 7px 0;
                    color:#1d4ed8;
                    font-size:{size}px;
                    font-weight:800;
                    line-height:1.4;
                ">
                    {title}
                </div>
                """
            )

            continue

        # Bullet
        bullet_match = re.match(
            r"^[-*•]\s+(.*)$",
            line
        )

        if bullet_match:

            bullet = bullet_match.group(1)

            bullet = re.sub(
                r"\*\*(.*?)\*\*",
                r"<b>\1</b>",
                bullet
            )

            rendered.append(
                f"""
                <div style="
                    color:#374151;
                    font-size:14px;
                    line-height:1.65;
                    margin:4px 0 4px 15px;
                ">
                    • {bullet}
                </div>
                """
            )

            continue

        # Bold text
        line = re.sub(
            r"\*\*(.*?)\*\*",
            r"<b style='color:#111827;'>\1</b>",
            line
        )

        rendered.append(
            f"""
            <div style="
                color:#374151;
                font-size:14px;
                line-height:1.7;
                margin:4px 0;
            ">
                {line}
            </div>
            """
        )

    return "".join(
        rendered
    )


# ==============================================================
# 14. GENAI FUNCTION
# ==============================================================

def generate_ai_explanation(_):

    with genai_output:

        clear_output(
            wait=True
        )

        try:

            # Current K-Means input
            values = np.array([
                [
                    float(heart_rate_input.value),
                    float(spo2_input.value),
                    float(glucose_input.value),
                    float(stress_input.value),
                    float(steps_input.value)
                ]
            ])

            scaled_values = (
                kmeans_scaler.transform(
                    values
                )
            )

            profile = int(
                kmeans_model.predict(
                    scaled_values
                )[0]
            )

            averages = (
                profile_summary.loc[
                    profile
                ]
            )

            prompt = f"""
You are a patient education assistant.

Explain the following dataset-derived monitoring profile
in simple, calm, patient-friendly language.

Assigned monitoring profile:
Profile {profile}

Entered monitoring values:
Heart rate = {heart_rate_input.value}
SpO2 = {spo2_input.value}
Blood glucose = {glucose_input.value}
Stress index = {stress_input.value}
Step count = {steps_input.value}

Profile {profile} averages:
Heart rate = {averages['heart_rate']:.2f}
SpO2 = {averages['spo2_level']:.2f}
Blood glucose = {averages['blood_glucose']:.2f}
Stress index = {averages['stress_level_index']:.2f}
Step count = {averages['step_count']:.2f}

Write a concise explanation using these sections:

### 1. What the Monitoring Profile Means

Explain that the profile is a statistical grouping
produced by K-Means from the dataset.

### 2. Values Relatively Close to the Profile Average

Compare the entered values with the profile averages.

### 3. Values That Differ More

Describe the larger numerical differences.

### 4. Why This Is a Statistical Comparison, Not a Diagnosis

Explain that comparison with a cluster average does
not diagnose disease or establish clinical risk.

### 5. Why Professional Medical Interpretation May Still Be Needed

Explain that healthcare professionals evaluate
individual context such as symptoms and medical history.

IMPORTANT RULES:

- Do not diagnose the person.
- Do not call the profile a disease.
- Do not call it a medical risk category.
- Do not say that a difference from an average means
  something is medically wrong.
- Refer to these as entered monitoring values.
- Do not make statements such as "this is safe"
  or "this is dangerous".
- Keep the language simple.
"""


            # --------------------------------------------------
            # Find API key
            # --------------------------------------------------

            api_key = os.environ.get(
                "GEMINI_API_KEY"
            )

            if not api_key:

                try:

                    from google.colab import userdata

                    api_key = userdata.get(
                        "GEMINI_API_KEY"
                    )

                except Exception:

                    pass


            if not api_key:

                display(
                    HTML(
                        card(

                            """
                            <div style="
                                color:#b45309;
                                font-size:20px;
                                font-weight:800;
                            ">
                                GenAI API Key Not Found
                            </div>

                            <div style="
                                color:#374151;
                                font-size:14px;
                                margin-top:9px;
                                line-height:1.6;
                            ">
                                Add GEMINI_API_KEY to your Colab
                                environment or Colab Secrets,
                                then click the button again.
                            </div>
                            """,

                            accent="#f59e0b"
                        )
                    )
                )

                return


            # --------------------------------------------------
            # Gemini
            # --------------------------------------------------

            from google import genai

            client = genai.Client(
                api_key=api_key
            )

            # Current documented models.
            model_candidates = [

                "gemini-3.5-flash-lite",

                "gemini-3.8-flash"

            ]

            response = None
            successful_model = None
            errors = []

            for model_name in model_candidates:

                try:

                    response = (
                        client
                        .models
                        .generate_content(
                            model=model_name,
                            contents=prompt
                        )
                    )

                    successful_model = (
                        model_name
                    )

                    break

                except Exception as exc:

                    errors.append(
                        f"{model_name}: {exc}"
                    )

            if response is None:

                raise RuntimeError(
                    "All configured Gemini models failed.\n\n"
                    + "\n".join(errors)
                )


            ai_text = (
                response.text
                or
                "No text was returned."
            )


            rendered = (
                markdown_to_html(
                    ai_text
                )
            )


            display(
                HTML(
                    card(

                        f"""
                        <div style="
                            color:#92400e;
                            font-size:11px;
                            font-weight:800;
                            text-transform:uppercase;
                            letter-spacing:.7px;
                            margin-bottom:10px;
                        ">
                            Generated by {successful_model}
                            · Profile {profile}
                        </div>

                        <div style="
                            color:#374151;
                            font-size:14px;
                        ">
                            {rendered}
                        </div>

                        <div style="
                            background:#fffbeb;
                            border:1px solid #fde68a;
                            border-radius:8px;
                            padding:10px 12px;
                            margin-top:16px;
                            color:#78350f;
                            font-size:12.5px;
                            line-height:1.6;
                        ">
                            This explanation is generated for
                            informational awareness. It is not
                            medical advice, diagnosis or treatment
                            guidance.
                        </div>
                        """,

                        accent="#f59e0b"
                    )
                )
            )


        except Exception as exc:

            display(
                HTML(
                    card(

                        f"""
                        <div style="
                            color:#b45309;
                            font-size:20px;
                            font-weight:800;
                        ">
                            Generative AI Temporarily Unavailable
                        </div>

                        <div style="
                            color:#374151;
                            font-size:14px;
                            line-height:1.6;
                            margin-top:9px;
                        ">
                            The K-Means monitoring profile was
                            computed successfully, but the
                            external Generative AI service did
                            not return a response.
                        </div>

                        <div style="
                            color:#6b7280;
                            font-size:12px;
                            line-height:1.5;
                            margin-top:10px;
                            word-break:break-word;
                        ">
                            Technical detail:<br>
                            {html.escape(str(exc))}
                        </div>
                        """,

                        accent="#f59e0b"
                    )
                )
            )


generate_ai_button.on_click(
    generate_ai_explanation
)


# ==============================================================
# 15. COMPLETE SYSTEM FLOW
# ==============================================================

display(
    HTML(

        card(

            section_title(
                "Complete System Flow",
                "End-to-end architecture of the project."
            )

            +

            """
            <div style="
                margin-top:15px;
                text-align:center;
                color:#111827;
                font-size:16px;
                font-weight:800;
                line-height:2;
            ">

                Sensor Data
                &nbsp;→&nbsp;
                Amazon S3
                &nbsp;→&nbsp;
                PySpark
                &nbsp;→&nbsp;
                Machine Learning
                &nbsp;→&nbsp;
                Monitoring Profile
                &nbsp;→&nbsp;
                Generative AI

            </div>

            <div style="
                background:#f9fafb;
                border:1px solid #e5e7eb;
                border-radius:8px;
                padding:10px 12px;
                margin-top:14px;
                color:#374151;
                font-size:12.5px;
                line-height:1.6;
            ">
                K-Means profiles are statistical groupings
                learned from the dataset. They are not medical
                diagnoses or validated clinical risk categories.
            </div>
            """,

            accent="#7c3aed"
        )

    )
)


# ==============================================================
# 16. FINAL STATUS
# ==============================================================

display(
    HTML(

        """
        <div style="
            background:#0f172a;
            border-radius:10px;
            padding:12px;
            margin-top:10px;
            text-align:center;
            font-family:Arial,sans-serif;
        ">

            <div style="
                color:#22c55e;
                font-size:14px;
                font-weight:800;
            ">
                ✓ SMART HEALTHCARE AI DASHBOARD READY
            </div>

            <div style="
                color:#cbd5e1;
                font-size:12px;
                margin-top:4px;
            ">
                K-Means + Random Forest + Generative AI
            </div>

        </div>
        """

    )
)



# ============================================================
# CLEAR OLD GENAI OUTPUT WHEN K-MEANS INPUT CHANGES
# ============================================================

def clear_old_genai(change):
    with genai_output:
        clear_output(wait=True)

        display(HTML("""
        <div style="
            background:#eff6ff;
            border:1px solid #bfdbfe;
            border-radius:8px;
            padding:10px;
            color:#1e3a8a;
            font-family:Arial,sans-serif;
            font-size:13px;
        ">
            Sensor values changed.
            Click <b>Generate AI Explanation</b> to create
            a new explanation for the current monitoring profile.
        </div>
        """))


heart_rate_input.observe(
    clear_old_genai,
    names="value"
)

spo2_input.observe(
    clear_old_genai,
    names="value"
)

glucose_input.observe(
    clear_old_genai,
    names="value"
)

stress_input.observe(
    clear_old_genai,
    names="value"
)

steps_input.observe(
    clear_old_genai,
    names="value"
)

# ==============================================================
# 17. CONSOLE VERIFICATION
# ==============================================================

print("\n" + "=" * 65)
print("SMART HEALTHCARE AI — FINAL DASHBOARD")
print("=" * 65)

print(f"Dataset records          : {len(df):,}")
print(f"K-Means profiles         : 12")
print(f"K-Means silhouette       : {kmeans_silhouette:.4f}")
print(f"RF training records      : {len(train_indices):,}")
print(f"RF held-out test records : {len(test_indices):,}")
print(f"RF test accuracy         : {rf_accuracy:.4f}")
print(f"RF macro F1              : {rf_macro_f1:.4f}")

print("=" * 65)
print("Dashboard loaded successfully.")
print("=" * 65)
````
